```python
import pandas as pd
import numpy as np
import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, MT5EncoderModel, get_scheduler
from torch.optim import AdamW
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, classification_report, confusion_matrix, roc_auc_score
from sklearn.model_selection import train_test_split
from collections import defaultdict
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import time
import gc


#Hyperparameters
TRAIN_BATCH_SIZE = 32
VALID_BATCH_SIZE = 32  
TEST_BATCH_SIZE = 32
EPOCHS = 12
MAX_LEN = 128
LR = 1e-5


# Load dataset
df = pd.read_csv("/kaggle/input/cse422/Bengali_Banglish_80K_Dataset.csv")
df = df[["Banglish", "Label"]].dropna().drop_duplicates().reset_index(drop=True)
df = df[df['Banglish'].str.strip() != '']
print(df['Label'].value_counts())


# Encode labels
label_encoder = LabelEncoder()
df['label'] = label_encoder.fit_transform(df['Label'])
num_classes = len(label_encoder.classes_)

#Dataset Split
train_texts, temp_texts, train_labels, temp_labels = train_test_split(df['Banglish'].tolist(), df['label'].tolist(), test_size=0.2, stratify=df['label'], random_state=42)
val_texts, test_texts, val_labels, test_labels = train_test_split(temp_texts, temp_labels, test_size=0.5, stratify=temp_labels, random_state=42)

#Dataset Class
class BanglaEmotionDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            self.texts[idx],
            padding='max_length',
            truncation=True,
            max_length=self.max_len,
            return_tensors="pt"
        )

        return {
            'input_ids': encoding['input_ids'].squeeze(),
            'attention_mask': encoding['attention_mask'].squeeze(),
            'label': torch.tensor(self.labels[idx], dtype=torch.long)
        }


def create_loaders(tokenizer):
    train_dataset = BanglaEmotionDataset(train_texts, train_labels, tokenizer, MAX_LEN)
    val_dataset = BanglaEmotionDataset(val_texts, val_labels, tokenizer, MAX_LEN)
    test_dataset = BanglaEmotionDataset(test_texts, test_labels, tokenizer, MAX_LEN)

    train_loader = DataLoader(train_dataset, batch_size=TRAIN_BATCH_SIZE, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=VALID_BATCH_SIZE, shuffle=False)
    test_loader = DataLoader(test_dataset, batch_size=TEST_BATCH_SIZE, shuffle=False)

    return train_loader, val_loader, test_loader



#Model Class for BanglaT5
class BanglaT5Classifier(nn.Module):
    def __init__(self, model_name, num_classes):
        super().__init__()
        self.encoder = AutoModelForSeq2SeqLM.from_pretrained(model_name).get_encoder()
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.encoder.config.hidden_size, num_classes)

    def forward(self, input_ids, attention_mask):
        encoder_outputs = self.encoder(input_ids=input_ids, attention_mask=attention_mask)
        pooled = encoder_outputs.last_hidden_state[:, 0]  # CLS token representation
        logits = self.classifier(self.dropout(pooled))
        return logits

#Model Class for MT5
class MT5Classifier(nn.Module):
    def __init__(self, model_name, num_classes):
        super().__init__()
        self.encoder = MT5EncoderModel.from_pretrained(model_name)
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(self.encoder.config.d_model, num_classes)

    def forward(self, input_ids, attention_mask):
        outputs = self.encoder(input_ids=input_ids, attention_mask=attention_mask)
        mask_exp = attention_mask.unsqueeze(-1).expand(outputs.last_hidden_state.size()).float()
        pooled = torch.sum(outputs.last_hidden_state * mask_exp, 1) / torch.clamp(mask_exp.sum(1), min=1e-9)
        return self.classifier(self.dropout(pooled))


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
loss_fn = nn.CrossEntropyLoss()

#Train Function
def train_model(train_loader, model, optimizer, loss_fn, scheduler=None):
    model.train()
    running_loss, correct_preds, total_samples = 0.0, 0, 0
    all_preds, all_labels  = [], []

    for batch in tqdm(train_loader, desc="Training"):
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['label'].to(device)

        outputs = model(input_ids, attention_mask)
        loss = loss_fn(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        if scheduler:
            scheduler.step()

        running_loss += loss.item()
        preds = torch.argmax(outputs, dim=1)
        correct_preds += torch.sum(preds == labels).item()
        total_samples += labels.size(0)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    avg_loss = running_loss / len(train_loader)
    accuracy = correct_preds / total_samples
    f1 = f1_score(all_labels, all_preds, average='weighted')
    return accuracy, f1, avg_loss



def eval_model(eval_loader, model, loss_fn):
    model.eval()
    all_preds = []
    all_labels = []
    all_probs = []
    total_loss = 0.0

    with torch.no_grad():
        for batch in tqdm(eval_loader, desc="Evaluating"):
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['label'].to(device)

            outputs = model(input_ids, attention_mask)
            loss = loss_fn(outputs, labels)
            total_loss += loss.item()

            probs = F.softmax(outputs, dim=1)
            preds = torch.argmax(probs, dim=1)

            all_probs.extend(probs.cpu().numpy())
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    accuracy = accuracy_score(all_labels, all_preds)
    # f1 = f1_score(all_labels, all_preds, average='weighted')
    # precision = precision_score(all_labels, all_preds, average='weighted')
    precision = precision_score(all_labels, all_preds, average='weighted', zero_division=0)
    f1 = f1_score(all_labels, all_preds, average='weighted', zero_division=0)
    recall = recall_score(all_labels, all_preds, average='weighted')
    avg_loss = total_loss / len(eval_loader)

    print(f"\nAccuracy: {accuracy:.4f}")
    print(f"F1 Score: {f1:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"Average Loss: {avg_loss:.4f}")

    # AUC ROC
    try:
        num_classes = len(set(all_labels))
        labels_bin = label_binarize(all_labels, classes=range(num_classes))
        auc_score = roc_auc_score(labels_bin, all_probs, average='weighted', multi_class='ovr')
        print(f"AUC ROC Score: {auc_score:.4f}")
    except ValueError:
        auc_score = float('nan')
        print("AUC ROC Score: Undefined (likely due to single class prediction)")

    print("\nClassification Report:\n")
    # print(classification_report(all_labels, all_preds, digits=4))
    print(classification_report(all_labels, all_preds, digits=4, zero_division=0))

    # Per-class accuracy and AUC
    print("Per-Class Accuracy and AUC:\n")
    cm = confusion_matrix(all_labels, all_preds)
    class_accuracies = cm.diagonal() / cm.sum(axis=1)
    for i, acc in enumerate(class_accuracies):
        class_name = label_encoder.inverse_transform([i])[0]
        correct = cm[i, i]
        total = cm.sum(axis=1)[i]
        try:
            class_auc = roc_auc_score(labels_bin[:, i], [prob[i] for prob in all_probs])
        except ValueError:
            class_auc = float('nan')
        print(f"Class {i} ({class_name}): {correct}/{total} correct ({acc:.4f} accuracy), AUC = {class_auc:.4f}")

    return accuracy, f1, avg_loss



# Plotting Loss and F1 Score over Epochs
def plot_metrics(history, model_name):
    epochs = range(1, len(history['train_loss']) + 1)

    plt.figure(figsize=(14, 5))

    # Plot Loss
    plt.subplot(1, 2, 1)
    plt.plot(epochs, history['train_loss'], label='Train Loss', marker='o')
    plt.plot(epochs, history['val_loss'], label='Validation Loss', marker='o')
    plt.title(f'{model_name} Loss over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    # Plot F1 Score
    plt.subplot(1, 2, 2)
    plt.plot(epochs, history['val_acc'], label='Validation Accuracy', marker='o')
    plt.plot(epochs, history['val_f1'], label='Validation F1 Score', marker='o')
    plt.title(f'{model_name} Accuracy/F1 over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Score')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()



# ---------------- BanglaT5 ----------------
print(f"\nTraining: BanglaT5\n" + "=" * 50)
tokenizer = AutoTokenizer.from_pretrained("csebuetnlp/banglat5")
train_loader, val_loader, test_loader = create_loaders(tokenizer)

banglat5_model = BanglaT5Classifier("csebuetnlp/banglat5", num_classes).to(device)
optimizer = AdamW(banglat5_model.parameters(), lr=LR, weight_decay=1e-4)
total_steps = len(train_loader) * EPOCHS
scheduler = get_scheduler("linear", optimizer=optimizer, num_warmup_steps=0, num_training_steps=total_steps)

start = time.time()
best_f1 = 0.0
history = defaultdict(list)

for epoch in range(1, EPOCHS + 1):
    print(f"\nEpoch {epoch}/{EPOCHS} [BanglaT5]")
    train_acc, train_f1, train_loss = train_model(train_loader, banglat5_model, optimizer, loss_fn, scheduler)
    val_acc, val_f1, val_loss = eval_model(val_loader, banglat5_model, loss_fn)

    history['train_acc'].append(train_acc)
    history['train_f1'].append(train_f1)
    history['train_loss'].append(train_loss)
    history['val_acc'].append(val_acc)
    history['val_f1'].append(val_f1)
    history['val_loss'].append(val_loss)

    if val_f1 > best_f1:
        torch.save(banglat5_model.state_dict(), "banglat5_model.bin")
        best_f1 = val_f1
        print("✅ Best BanglaT5 model saved!")

end = time.time()
print(f"⏱️ Total BanglaT5 training & evaluation time: {end - start:.2f} seconds")

print("\nTesting Best BanglaT5 Model...")
banglat5_model.load_state_dict(torch.load("banglat5_model.bin", weights_only=True))

start = time.time()
test_acc, test_f1, _ = eval_model(test_loader, banglat5_model, loss_fn)
end = time.time()
print(f"🧪 BanglaT5 test-set evaluation time: {end - start:.2f} seconds")
plot_metrics(history, "BanglaT5")


del banglat5_model
torch.cuda.empty_cache()
gc.collect()

# ---------------- MT5 ----------------
print(f"\nTraining: MT5\n" + "=" * 50)
tokenizer = AutoTokenizer.from_pretrained("google/mt5-base")
train_loader, val_loader, test_loader = create_loaders(tokenizer)


mt5_model = MT5Classifier("google/mt5-base", num_classes).to(device)
optimizer = AdamW(mt5_model.parameters(), lr=LR, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10)

start = time.time()
best_f1 = 0.0
history = defaultdict(list)

for epoch in range(1, EPOCHS + 1):
    print(f"\nEpoch {epoch}/{EPOCHS} [MT5]")
    train_acc, train_f1, train_loss = train_model(train_loader, mt5_model, optimizer, loss_fn, scheduler)
    val_acc, val_f1, val_loss = eval_model(val_loader, mt5_model, loss_fn)

    history['train_acc'].append(train_acc)
    history['train_f1'].append(train_f1)
    history['train_loss'].append(train_loss)
    history['val_acc'].append(val_acc)
    history['val_f1'].append(val_f1)
    history['val_loss'].append(val_loss)

    if val_f1 > best_f1:
        torch.save(mt5_model.state_dict(), "mt5_model.bin")
        best_f1 = val_f1
        print("✅ Best MT5 model saved!")

end = time.time()
print(f"⏱️ Total MT5 training & evaluation time: {end - start:.2f} seconds")

print("\nTesting Best MT5 Model...")
mt5_model.load_state_dict(torch.load("mt5_model.bin", weights_only=True))
start = time.time()
test_acc, test_f1, _ = eval_model(test_loader, mt5_model, loss_fn)
end = time.time()
print(f"🧪 MT5 test-set evaluation time: {end - start:.2f} seconds")
print(f"📊 Test Accuracy: {test_acc:.4f}, F1 Score: {test_f1:.4f}")
plot_metrics(history, "MT5")
del mt5_model
torch.cuda.empty_cache()
gc.collect()
```
